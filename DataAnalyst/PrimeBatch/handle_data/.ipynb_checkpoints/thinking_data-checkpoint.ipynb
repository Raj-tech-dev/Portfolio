```python
#cleaning the data
def clean_data(data):
    text_to_num = {"one": 1, "two": 2,"three": 3, "four": 4,"five": 5}
    cleaned_data = []
    unique_data = set()

    
    for user in data:
        raw_rating = str(user["rating"]).strip().lower() #cleanining white space and change in lower case
        if(raw_rating in text_to_num):
            raw_rating = text_to_num[raw_rating]
        user["rating"] = raw_rating

        #handle with missing vlaue
        raw_age = user.get("age")
        if(raw_age == None):
            user["age"] = None

        #remove duplication data
        if(user["name"].strip() in unique_data):
            continue
        unique_data.add(user["name"])
        cleaned_data.append(user)
    return cleaned_data
        


        
data = clean_data(data)
```

```python
# find average rating and <3 rating 
def data_insights(data):
    #average rating 
    tot_rating = 0
    for user in data:
        tot_rating +=float( user["rating"])
        
```
