```python
# pandas practice.........
import pandas as pd
import numpy as np
```

```python
 
# from scaler value create series
s= pd.Series([1,2,3,4,5,6,7,8] , index=("a", "b", "c", "d", "e", "f", "g", "h"))
print(s)
```

```python
 # from dictionary create series
info = { "name": "raj", "kaam": "read", "goal": "fly"}
dic = pd.Series(info)
print(dic)
    
```

```python
# form ndarry craeye series
import numpy as np
nd = np.array([1,2,3,4,5])
s=pd.Series(nd)
print(s)
```

```python
s = pd.Series([1,2,3,4,5,6,7,8,9,0],index=["a", "b", "c", "d", "e", "f", "g", "h", "i", "j"])
print(s)
print(type(s))
print(s.index)
```

```python
#Indexing of series
s = pd.Series([1,2,3,4,5,6,7,8,9,0])
print(s[8])
print(s.index)
            
```
